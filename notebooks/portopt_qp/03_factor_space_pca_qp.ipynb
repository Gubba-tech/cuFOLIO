# Factor-Space PCA QP

PCA data is deterministic and synthetic; constraints act on stock exposure `V @ z`.

```python
import numpy as np
from cufolio import QPParameters, build_pca_factor_qp_data
backend = "osqp"  # switch explicitly to "cuopt" on a configured GPU host
stock_returns = np.random.default_rng(8).normal(0.002, 0.01, size=(60, 13))
data = build_pca_factor_qp_data(stock_returns, n_components=3, center=False)
params = QPParameters(mapping_mode="factor_space", V=data.stock_mapping, objective="max_sharpe", w_min=-0.08, w_max=0.08, short_budget=0.2, lambda_l1=0.01, lambda_l2=0.01, backend=backend)
print(data.factor_mean.shape, data.stock_mapping.shape, params.mapping_mode)
```
