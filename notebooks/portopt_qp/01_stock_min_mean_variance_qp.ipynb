# Stock Minimum- and Mean-Variance QP

Synthetic stock data keeps this notebook short and reproducible.

```python
import numpy as np
from cufolio.qp_backend import solve_compiled_qp_osqp, solve_compiled_qp_cuopt
from cufolio.qp_formulations import compile_portfolio_qp
from cufolio.qp_parameters import QPParameters
backend = "osqp"  # switch explicitly to "cuopt" on a configured GPU host
rng = np.random.default_rng(4)
A = rng.normal(size=(8, 8))
returns = {"mean": rng.normal(0.03, 0.004, 8), "covariance": A.T @ A / 8 + 0.02 * np.eye(8)}
params = QPParameters(objective="mean_variance", risk_aversion=2.0, backend=backend)
compiled = compile_portfolio_qp(returns, params)
solution = solve_compiled_qp_osqp(compiled) if backend == "osqp" else solve_compiled_qp_cuopt(compiled)
print(solution.status, compiled.recover_stock_weights(solution.x).sum())
```
