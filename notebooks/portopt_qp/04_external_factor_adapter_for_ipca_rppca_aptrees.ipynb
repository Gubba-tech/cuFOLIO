# External Factor Adapter

Externally computed RP-PCA, IPCA, or AP-Trees factor returns and mapping `V` can enter through the same adapter. Full estimator replication is outside this notebook.

```python
import numpy as np
from cufolio import QPParameters, build_external_factor_qp_data
backend = "osqp"  # switch explicitly to "cuopt" on a configured GPU host
rng = np.random.default_rng(9)
factor_returns = rng.normal(0.01, 0.02, size=(50, 3))
V = rng.uniform(0.05, 1.0, size=(12, 3)); V /= V.sum(axis=0, keepdims=True)
data = build_external_factor_qp_data(factor_returns, V, model_name="external")
params = QPParameters(mapping_mode="factor_space", V=V, objective="mean_variance", backend=backend)
print(data.factor_mean.shape, data.stock_mapping.shape, params.mapping_mode)
```
